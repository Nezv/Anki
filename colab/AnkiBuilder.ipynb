# Anki Builder

Builds an `.apkg` from the batch files in [github.com/Nezv/Anki](https://github.com/Nezv/Anki) and saves it to `MyDrive/Anki/`.

**Usage:** Runtime ▸ Run all. Then open the `.apkg` from Google Drive on any device with Anki installed.

Re-importing is safe: note GUIDs are stable, so existing cards get **updated**, not duplicated — your review history is preserved.

```python
#@title 3. Build the .apkg into Drive
!mkdir -p '/content/drive/MyDrive/Anki'
!python /content/Anki/tools/build_apkg.py --out '/content/drive/MyDrive/Anki'
```

### Optional: paste a batch directly

If you have new note strings that aren't in the repo yet, paste them into the cell below and run it — it builds only that batch.

```python
#@title 4. (Optional) Build from pasted strings
batch = '''\
#notetype: hanzi
#deck: 汉语
Hanzi:好|Pinyin:hǎo|Meaning:good|Words:你好,好看|WordPinyin:nǐ hǎo,hǎo kàn|WordMeanings:hello,good-looking|Radicals:女 子|Sentence:你好吗？|SentencePinyin:nǐ hǎo ma|SentenceMeaning:How are you?
'''

import pathlib
tmp = pathlib.Path('/content/pasted_batch')
tmp.mkdir(exist_ok=True)
(tmp / 'pasted.txt').write_text(batch, encoding='utf-8')
!python /content/Anki/tools/build_apkg.py --batches /content/pasted_batch --out '/content/drive/MyDrive/Anki'
```
